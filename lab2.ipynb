```python
import random

# Part A – Machine Initialization
machines = ['A', 'B', 'C']
M = len(machines)
N_play = 10
epsilon = 0.2

# Part B – Initialize knowledge structures (N: counts, R: sum of rewards, Q: values)
N = {m: 0 for m in machines}
R = {m: 0.0 for m in machines}
Q = {m: 0.0 for m in machines}
total_reward_received = 0.0

print("=== Starting e-Greedy Multi-Armed Bandit (10 Plays) ===\n")

for play in range(1, N_play + 1):
    print(f"--- Play {play} ---")

    # Part C – ε-Greedy Decision
    r = random.random()
    if r < epsilon:
        # Exploration
        selected_machine = random.choice(machines)
        action_type = "Exploration"
    else:
        # Exploitation
        # Find max Q-value
        max_q = max(Q.values())
        # Tie-breaking randomly among machines with the max Q-value
        best_machines = [m for m in machines if Q[m] == max_q]
        selected_machine = random.choice(best_machines)
        action_type = "Exploitation"

    print(f"Selected Machine: {selected_machine} ({action_type})")

    # Part D – Reward Collection
    while True:
        try:
            reward = float(input(f"Play with machine {selected_machine}\nEnter reward: "))
            break
        except ValueError:
            print("Please enter a valid numerical value for reward.")

    # Part E – Q-Value Update
    # 1. Update action count
    N[selected_machine] += 1
    # 2. Accumulate reward
    R[selected_machine] += reward
    total_reward_received += reward

    # 3. Update Q-value using incremental update formula:
    # Q_new = Q_old + (1/N) * (Reward - Q_old)
    step_size = 1.0 / N[selected_machine]
    Q[selected_machine] = Q[selected_machine] + step_size * (reward - Q[selected_machine])

    # Print current status
    print(f"Updated N: {N}")
    print(f"Updated Q: {Q}\n")

# Part F – Display Results
print("==================================================")
print("FINAL RESULTS & ANALYSIS")
print("==================================================")
print(f"{'Machine':<10}{'Plays (N)':<15}{'Total Reward (R)':<20}{'Estimated Q-value':<20}")
for m in machines:
    print(f"{m:<10}{N[m]:<15}{R[m]:<20.2f}{Q[m]:<20.4f}")
print("--------------------------------------------------")
print(f"Total reward received by the player: {total_reward_received:.2f}")
```
