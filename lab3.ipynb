```python
import random

# Part A: Machine Initialization
machines = ['A', 'B', 'C']
M = len(machines)
N_play = 10
epsilon = 0.2

# Part B: Create knowledge tables (Initialization of N, R, and Q)
# N: Participation count for each machine
N = {m: 0 for m in machines}
# R: List of rewards received for each machine
R = {m: [] for m in machines}
# Q: Estimated reward (Q-value) for each machine
Q = {m: 0.0 for m in machines}

print("Initial State:")
print(f"N (Counts): {N}")
print(f"Q (Estimates): {Q}\n")

# Tracking total reward
total_reward = 0

for play in range(1, N_play + 1):
    print(f"--- Play {play} ---")

    # Part C: Modified epsilon-Greedy Decision
    # Determine the machine(s) with the highest Q-value
    max_q = max(Q.values())
    best_machines = [m for m, val in Q.items() if val == max_q]

    # In case of ties, choose one of the best arbitrarily
    greedy_machine = random.choice(best_machines)

    r = random.random()
    if r < epsilon:
        # Exploration: Select randomly except the one with the highest Q-value
        non_greedy_machines = [m for m in machines if m != greedy_machine]
        # Fallback if all have same Q-value, just pick from remaining
        selected_machine = random.choice(non_greedy_machines if non_greedy_machines else machines)
        strategy = "Exploration"
    else:
        # Exploitation
        selected_machine = greedy_machine
        strategy = "Exploitation"

    print(f"Strategy: {strategy} -> Selected Machine: {selected_machine}")

    # Part D: Reward Collection
    while True:
        try:
            reward_input = input(f"Play with machine {selected_machine}. Enter reward: ")
            reward = float(reward_input)
            break
        except ValueError:
            print("Please enter a valid numerical reward.")

    # Part E: Q-Value Update
    N[selected_machine] += 1
    R[selected_machine].append(reward)
    total_reward += reward

    # Incremental update formula: Q_new = Q_old + (1 / N) * (Reward - Q_old)
    n_val = N[selected_machine]
    q_old = Q[selected_machine]
    Q[selected_machine] = q_old + (1.0 / n_val) * (reward - q_old)

    # Display state update
    print(f"Updated N for {selected_machine}: {N[selected_machine]}")
    print(f"Updated Q for {selected_machine}: {Q[selected_machine]:.4f}\n")

# Part F: Display Results
print("=================== FINAL RESULTS ===================")
print(f"Participation Count (N): {N}")
print(f"All Received Rewards (R): {R}")
print(f"Final Estimated Q-values (Q): {Q}")
print(f"Total Reward Obtained: {total_reward}")
print("=====================================================")
```
