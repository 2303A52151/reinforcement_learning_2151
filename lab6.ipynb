### Step 2: Implement the Random Agent Runner Function
This function initializes the environment, runs the random agent for a specified number of episodes, selects actions randomly, accumulates rewards, and prints the performance metrics.

```python
import gymnasium as gym
import numpy as np

def run_random_agent(env_name, num_episodes=5):
    print(f"=== Running Random Agent in {env_name} ===\n")
    env = gym.make(env_name)

    # Display basic environment information
    print(f"Action Space: {env.action_space}")
    print(f"Observation Space: {env.observation_space}\n")

    total_rewards_all_episodes = []

    for episode in range(1, num_episodes + 1):
        state, info = env.reset()
        done = False
        episode_reward = 0
        step_count = 0

        while not done:
            # Select an action randomly from the environment's action space
            action = env.action_space.sample()

            # Execute the action
            next_state, reward, terminated, truncated, info = env.step(action)

            episode_reward += reward
            step_count += 1

            # Environment is finished if either terminated or truncated
            done = terminated or truncated
            state = next_state

        total_rewards_all_episodes.append(episode_reward)
        print(f"Episode {episode}: Total Steps = {step_count}, Total Reward = {episode_reward}")

    env.close()

    # Calculate overall metrics
    avg_reward = np.mean(total_rewards_all_episodes)
    print(f"\nSummary for {env_name}:")
    print(f"Total Reward over {num_episodes} episodes: {sum(total_rewards_all_episodes)}")
    print(f"Average Reward per episode: {avg_reward:.2f}\n")
```

### Step 3: Run the Experiments
Now, we run our random agent on both `CliffWalking-v0` and `Taxi-v3` environments.

```python
# Run CliffWalking-v1 environment
run_random_agent("CliffWalking-v1", num_episodes=5)
```

```python
# Run Taxi-v4 environment
run_random_agent("Taxi-v4", num_episodes=5)
```
